# Ejercicio 4 - Estadística descriptiva

Análisis de los datos generados en el juego del ahorcado utilizando pandas y matplotlib.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Cargar datos
words = pd.read_csv("C:/Users/aangeles/OneDrive - Future Space/Practicas_Antonio/HangmanPython/data/words.csv", header=None, names=["word"])

rounds = pd.read_csv("C:/Users/aangeles/OneDrive - Future Space/Practicas_Antonio/HangmanPython/data/rounds_in_games.csv")
rounds.columns = rounds.columns.str.strip()

games = pd.read_csv("C:/Users/aangeles/OneDrive - Future Space/Practicas_Antonio/HangmanPython/data/games.csv")
games.columns = games.columns.str.strip()

# Convertir columnas a número
games["final_score"] = pd.to_numeric(games["final_score"], errors="coerce")
rounds["user_trys"] = pd.to_numeric(rounds["user_trys"], errors="coerce")

print("Datos cargados correctamente")
```

## Distribución por número de letras

```python
words["length"] = words["word"].apply(len)

conteo = words["length"].value_counts().sort_index()

conteo.plot(kind="bar")

plt.title("Distribución por número de letras")
plt.xlabel("Número de letras")
plt.ylabel("Cantidad")

plt.show()
```

## Análisis de rondas

```python
usuarios_unicos = rounds["username"].nunique()

print("Número de usuarios únicos:", usuarios_unicos)
```

```python
ganadas = rounds[rounds["victory"] == True]

top5 = ganadas["username"].value_counts().head(5)

print("Top 5 usuarios con más rondas ganadas:")
print(top5)
```

## Análisis de partidas

```python
top_total = games.groupby("username")["final_score"].sum().sort_values(ascending=False).head(5)

print("Top 5 usuarios con mayor puntuación acumulada:")
print(top_total)
```

```python
top_media = games.groupby("username")["final_score"].mean().sort_values(ascending=False).head(5)

print("Top 5 usuarios con mayor media de puntuación:")
print(top_media)
```

```python
games["final_score"].plot(kind="hist", bins=4)

plt.title("Distribución de puntuaciones finales")
plt.xlabel("Puntuación")
plt.ylabel("Frecuencia")

plt.show()
```
